Cell 1: define the schemas up front

```python

customers_schema = "Customer_ID STRING, Customer_Name STRING, City STRING, Mobile_Number STRING, KYC_Status STRING"
accounts_schema  = "Account_ID STRING, Customer_ID STRING, Account_Type STRING, Balance STRING, Status STRING"
txn_schema       = "Transaction_ID STRING, Account_ID STRING, Transaction_Date STRING, Amount STRING, Type STRING, Status STRING"
raw = "/Volumes/banking/dq/raw"
```

1. Load the four supplied CSV datasets into Azure Databricks. Display each dataset’s schema and total row count.

```python
df_customers = spark.read.csv("/Volumes/banking/dq/raw/customers.csv", header=True, schema=customers_schema)
display(df_customers)
df_customers.write.mode("overwrite").saveAsTable("banking.dq.bronze_customers")



df_accounts = spark.read.csv("/Volumes/banking/dq/raw/accounts.csv", header=True, schema=accounts_schema)
display(df_accounts)
df_accounts.write.mode("overwrite").saveAsTable("banking.dq.bronze_accounts")


df_source = spark.read.csv("/Volumes/banking/dq/raw/source_transactions.csv", header=True, schema=txn_schema)
display(df_source)
df_source.write.mode("overwrite").saveAsTable("banking.dq.bronze_source_transactions")

df_target= spark.read.csv("/Volumes/banking/dq/raw/target_transactions.csv", header=True, schema=txn_schema)
display(df_target)
df_target.write.mode("overwrite").saveAsTable("banking.dq.bronze_target_transactions")

df_customers.printSchema()
print("customers rows:", df_customers.count())

df_accounts.printSchema()
print("accounts rows:", df_accounts.count())

df_source.printSchema()
print("source rows:", df_source.count())

df_target.printSchema()
print("target rows:", df_target.count())
```
